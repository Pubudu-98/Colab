# ATM Operation in Python

```python
account_balance = 1000

def check_balance():
  print('Your Current Balance is', account_balance)
```

```python
def deposit(amount):
  global account_balance

  if amount > 0:
    account_balance += amount
    print('Deposit: LKR', amount)
    check_balance()
  else:
    print('Invalid Amount!')
```

```python
def withdraw(amount):
  global account_balance

  if amount > 0 and amount <= account_balance:
    account_balance -= amount
    print('Withdrawal: LKR', amount)
    check_balance()
  else:
    print('Invalid Amount!')
```

```python
def atm_menu():
    while True:
        print('----Automated Teller Machine----')
        print('1. Check Balance')
        print('2. Deposit')
        print('3. Withdraw')
        print('4. Exit')
        print('--------------------------------')

        choice = input('Enter Your Selection: ')

        if choice == '1':
            check_balance()
        elif choice == '2':
            try:
                amount = float(input('Enter Deposit Amount: LKR'))
                deposit(amount)
            except ValueError:
                print('Invalid Amount!')
        elif choice == '3':
          try:
              amount = float(input('Enter Withdrawal Amount: LKR'))
              withdraw(amount)
          except ValueError:
            print('Invalid Amount!')
        elif choice == '4':
          print('Thank you for using our ATM, Have a good day!!!')

          break
```

```python
atm_menu()
```

# **Automatic Climate Control System**

```python
def climate_control_system(current_temperature, target_temperature=22):
  print('Current Temperature:', current_temperature, '°C')
  print('Target Temperature:', target_temperature, '°C')

  if current_temperature > target_temperature + 2:
    print('Activating Cooling System')
  elif current_temperature < target_temperature - 2:
    print('Activating Heating Mode')
  else:
    print('Climate is Comfort!')
```

**# Test Cases**

```python
print("--- Testing with target temperature 22°C ---")
climate_control_system(current_temperature=28) # Should activate cooling
climate_control_system(current_temperature=24) # Should activate cooling
climate_control_system(current_temperature=22) # Should be comfortable
climate_control_system(current_temperature=20) # Should be comfortable
climate_control_system(current_temperature=18) # Should activate heating
climate_control_system(current_temperature=15) # Should activate heating
```
